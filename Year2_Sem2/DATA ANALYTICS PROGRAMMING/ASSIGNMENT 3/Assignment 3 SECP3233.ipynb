# Rainfall Data Analysis Assignment (100 Marks, Due 6th July 2025)

You are provided with a rainfall dataset recording monthly measurements from various weather stations across different regions. The dataset includes rainfall (mm), number of rainy days, maximum and minimum temperatures, and whether the station is in an urban or rural area.

Dataset columns:

- `Region`: Name of the region (e.g., North, South, East, West)
- `Year`: Year of measurement
- `Month`: Month (e.g., Jan, Feb, ..., Dec)
- `Station`: Weather station name/ID (e.g., StationA, StationB)
- `Rainfall_mm`: Rainfall amount in millimeters
- `RainyDays`: Number of days in the month with rain
- `MaxTemp_C`: Maximum temperature recorded in that month (°C)
- `MinTemp_C`: Minimum temperature recorded in that month (°C)
- `UrbanRural`: "Urban" or "Rural"

## Part 1: Data Exploration ( 15 Marks)

1. Load the data into a pandas DataFrame. (2 Marks)
2. Display the first 5 rows of the DataFrame. (2 Marks)
3. Check for and handle any missing values. (6 Marks)
4. Provide summary statistics for all numeric columns. (5 Marks)

```python
#Write your code here
import pandas as pd
df = pd.read_csv('Rainfall.csv', header = None)
```

```python
#there is a problem in the dataset .csv as all datas are packed in one column
#so we need to split them manually
df = df[0].str.split(",", expand = True)
#expand = True is to expand it into dataframe columns

#reset again the column names follow the question given
df.columns = ['Region', 'Year', 'Month', 'Station', 'Rainfall_mm', 'RainyDays', 'MaxTemp_C', 'MinTemp_C', 'UrbanRural']

#remove the original unneat headers row
df = df[1:].reset_index(drop=True)

df.head()
```

```python
print("Missing value in each column:")
print(df.isnull().sum())

#if have missing value
df = df.dropna()

print("\nAfter Cleaning:")
print(df.isnull().sum())
```

```python
#convert columns from string to float/int
df['Year'] = df['Year'].astype(int)
df['Rainfall_mm'] = df['Rainfall_mm'].astype(float)
df['RainyDays'] = df['RainyDays'].astype(int)
df['MaxTemp_C'] = df['MaxTemp_C'].astype(float)
df['MinTemp_C'] = df['MinTemp_C'].astype(float)

df.describe()
```

## Part 3: Data Visualization (matplotlib) (40 Marks)

1. **Bar Chart:**  (8 Marks)
   - Plot total rainfall by region, colored by Urban/Rural.
   - Add value labels, gridlines, axis labels, title, and adjust bar width for clarity.



```python
import matplotlib.pyplot as plt

total_rainfall = df.groupby(['Region', 'UrbanRural'])['Rainfall_mm'].sum().reset_index()

#X-axis labels
x_labels = [f"{r} ({u})" for r, u in zip(total_rainfall['Region'], total_rainfall['UrbanRural'])]

# Colors by UrbanRural
colors = {'Urban': 'r', 'Rural': 'b'}
bar_colors = [colors[u] for u in total_rainfall['UrbanRural']]

# Step 2: Use lecture-style plot
fig, axes = plt.subplots(1, 1, figsize=(10, 6))  # only one axis

# Bar plot
axes.bar(x_labels, total_rainfall['Rainfall_mm'], color=bar_colors, alpha=0.8)

# Add labels and grid
axes.set_title("Total Rainfall by Region (Urban vs Rural)", fontsize=13)
axes.set_ylabel("Rainfall (mm)")
axes.set_xlabel("Region (Urban/Rural)")
axes.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels
for i, val in enumerate(total_rainfall['Rainfall_mm']):
    axes.text(i, val + 2, f"{val:.1f}", ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

2. **Grouped Bar Chart:**  (6 Marks)
   - Show average monthly rainfall for each station.
   - Use different colors for each station, include a legend, rotate x-tick labels, and increase figure size for readability.



```python
#Calculate average monthly rainfall for each station
rainfall_avg = df.groupby(['Month', 'Station'])['Rainfall_mm'].mean().unstack()

fig, ax = plt.subplots(figsize=(12, 6)) 

rainfall_avg.plot(kind='bar', ax=ax)

ax.set_title("Average Monthly Rainfall by Station", fontsize=14)
ax.set_ylabel("Rainfall (mm)")
ax.set_xlabel("Month")
ax.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=0)  # Rotate x-tick labels for readability
ax.legend(title="Station")

plt.tight_layout()
plt.show()


```

3. **Line Plot:**  (6 Marks)
   - Show the trend of rainfall over months for at least two selected stations.
   - Use markers, dashed lines, highlight the month with maximum rainfall, and format the x-axis for months.



```python
import matplotlib.pyplot as plt

# Choose two stations
stations = ['StationA', 'StationB']

# Filter and compute monthly average rainfall
monthly = df[df['Station'].isin(stations)].groupby(['Month', 'Station'])['Rainfall_mm'].mean().unstack()

# Order months properly (optional but good)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly = monthly.reindex(months)

# Plot
fig, ax = plt.subplots(figsize=(10, 5))
monthly.plot(ax=ax, marker='o', linestyle='--')

# Highlight max rainfall month (just once, not per station)
max_month = monthly.max(axis=1).idxmax()
max_value = monthly.max(axis=1).max()
ax.axvline(max_month, color='gray', linestyle=':', alpha=0.5)
ax.text(max_month, max_value + 2, f"Max: {max_value:.1f} mm", ha='center', fontsize=9)

# Styling
ax.set_title("Rainfall Trend Over Months (Two Stations)")
ax.set_ylabel("Rainfall (mm)")
ax.set_xlabel("Month")
ax.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=0)
ax.legend(title="Station")

plt.tight_layout()
plt.show()
```

4. **Boxplot:**  (6 Marks)
   - Display rainfall distributions by region and by Urban/Rural.
   - Use custom box colors, highlight outliers, and set y-axis limits.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up figure and axes
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Custom color palette
palette = {'North': '#87CEEB', 'South': '#FFA07A', 'East': '#98FB98', 'West': '#DDA0DD'}
urban_palette = {'Urban': '#6495ED', 'Rural': '#FF7F50'}

# --- Boxplot by Region ---
sns.boxplot(data=df, x='Region', y='Rainfall_mm', hue='Region', palette=palette, ax=axes[0],
            flierprops=dict(marker='o', markerfacecolor='red', markersize=6))  # Outliers style
axes[0].set_title("Rainfall Distribution by Region")
axes[0].set_ylabel("Rainfall (mm)")
axes[0].set_ylim(0, df['Rainfall_mm'].max() + 20)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# --- Boxplot by Urban/Rural ---
sns.boxplot(data=df, x='UrbanRural', y='Rainfall_mm', hue='UrbanRural', palette=urban_palette, ax=axes[1],
            flierprops=dict(marker='x', markerfacecolor='darkred', markersize=6))  # Outliers style
axes[1].set_title("Rainfall Distribution by Urban/Rural")
axes[1].set_ylabel("Rainfall (mm)")
axes[1].set_ylim(0, df['Rainfall_mm'].max() + 20)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
