#### 1. Dataset Description

##### Cars2025 Dataset

This dataset contains the actual transactional records of all cars registered with the Road Transport Department (JPJ) since 2000, excluding any personal identifiable data. Under Malaysian law, all vehicles must be registered with JPJ and be assigned a registration plate number before they can be used on the road. 

While there has been cleaning and validation conducted prior to publication, no other amendment or aggregation has been performed. Each row in the dataset represents one registration transaction. This is the most granular form of the data.

Attributes of the Car2025 dataset:

```python
import pandas as pd

carData = {'Description': ['Car registered date ', 'Type of vehicle', 'Car brand ', 'Car model name ', 'Fuel used by the car ', 'Colour of the car ','State where the car was registered ']} 

frame = pd.DataFrame(carData, columns = ['Description'],
                    index = ['date_reg ', 'type ', 'maker ', 'model ', 'fuel ', 'colour ', 'state '])
frame.index.name = 'Attributes'
frame

frame.style.set_properties(**{'text-align': 'left'})
```

##### FuelConsumption Dataset

The "FuelComsumption Dataset" contains comprehensive information on various vehicles. It includes details such as make, model, vehicle class, engine size, cylinder count, transmission type, and fuel type of those vehicles. Additionally, the dataset provides ranges for fuel consumption and CO2 emissions, offering insights into the environmental impact of each vehicle. The dataset encompasses a wide range of vehicle types, from compact to mid-size, full size, and includes both conventional and high-performance models. With this information, we can study trends in vehicle characteristics, fuel efficiency, and emissions. This dataset serves as a valuable resource for understanding the automotive landscape and informing discussions on environmental sustainability and transportation policies. 

```python
import pandas as pd


fuelData = {'Description': ['Year of manufacture for vehicle', 'Vehicle brand', 'Specific model of vehicle', 'Vehicle class based on their size (e.g. compact, mid-size)', 'Volume of engine in litres', 'Number of engine cylinders','Type of transmission system (e.g. A4(manual) or M5(auto))','Type of fuel used (e.g. X, Z)','Amount of fuel consumed by vehicle','Amount of CO2 emitted by vehicle']} 

frame = pd.DataFrame(fuelData, columns = ['Description'],
                    index = ['MODELYEAR', 'MAKE', 'MODEL', 'VEHICLECLASS', 'ENGINECLASS', 'CYLINDERS', 'TRANSMISSION','FUELTYPE','FUELCONSUMPTION','COEMISSIONS'])
frame.index.name = 'Attributes'
frame

frame.style.set_properties(**{'text-align': 'left'})
```
