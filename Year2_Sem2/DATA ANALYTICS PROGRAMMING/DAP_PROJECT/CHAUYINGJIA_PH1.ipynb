#### 1. Dataset Description

##### Cars2025 Dataset

This dataset contains the actual transactional records of all cars registered with the Road Transport Department (JPJ) since 2000, excluding any personal identifiable data. Under Malaysian law, all vehicles must be registered with JPJ and be assigned a registration plate number before they can be used on the road. 

While there has been cleaning and validation conducted prior to publication, no other amendment or aggregation has been performed. Each row in the dataset represents one registration transaction. This is the most granular form of the data.

#### 3. Data Cleaning and Preparation 

1. Import required libraries: 

```python
import pandas as pd
import re
```

2. Load both datasets 

```python
fuel_df = pd.read_csv('FuelConsumption.csv')     
cars_df = pd.read_csv('cars_2025.csv')
```

3. Preview the data info, description and the first five rows for fuelConsumption dataset

4. Check different columns name to standardize before merging 

```python
print("Cars2025 columns:", cars_df.columns.tolist())
print("FuelConsumption columns:", fuel_df.columns.tolist())
```

5. Check again for their missing values

```python
print(cars_df.isnull().sum())
print()
print(fuel_df.isnull().sum())
```

6. Remove duplicate rows

```python
cars_df = cars_df.drop_duplicates()
fuel_df = fuel_df.drop_duplicates()
```

7. Standardizing

```python
#Remove spaces, symbols and change string to lowercase
cars_df.columns = cars_df.columns.str.strip().str.lower().str.replace(' ', '_')
fuel_df.columns = fuel_df.columns.str.strip().str.lower().str.replace(' ', '_')

#rename the maker to make to standardize
cars_df.rename(columns={'maker': 'make'}, inplace=True)

#column values (e.g., maker, model, MAKE, MODEL) by converting to lowercase and stripping spaces 
cars_df['make'] = cars_df['make'].str.strip().str.upper()
cars_df['model'] = cars_df['model'].str.strip().str.upper()
cars_df['fuel'] = cars_df['fuel'].str.lower().str.strip()

fuel_df['make'] = fuel_df['make'].str.strip().str.upper()
fuel_df['model'] = fuel_df['model'].str.strip().str.upper()
fuel_df['fueltype'] = fuel_df['fueltype'].str.lower().str.strip()
```

8. Remove outliers

```python
#Remove outliers from fuelConsumption data
fuel_cleaned = fuel_df[
    (fuel_df['fuelconsumption_comb'] < 30) &
    (fuel_df['co2emissions'] < 600)
]

print(f"Cleaned cars_2025 shape: {cars_df.shape}")
print(f"Cleaned fuel_consumption shape: {fuel_cleaned.shape}")
```

9. Merged the two datasets

```python
# Merge datasets on make and model to match registrations with emissions info
merged_data = pd.merge(cars_df,fuel_cleaned,
                        on=['make', 'model'],
                        how='inner'
)

# Show the shape of the merged dataset
print(f"Merged data shape: {merged_data.shape}")

#Easy way to check on their column
print("merged_data columns:")
print(merged_data.columns.tolist())
```

```python
# Show the first 5 rows to see all columns combined
merged_data.info()
merged_data.describe()
merged_data.head()

import os

if os.path.exists("cleaned_data.csv"):
    os.remove("cleaned_data.csv")

    merged_data.to_csv('merged_data.csv', index=False)
```

#### 4. Data Aggregation and Group Operations 

1. Average CO2 Emissions by Fuel Type 

```python
merged_data['fueltype'] = merged_data['fueltype'].replace({
    'x': 'Petrol',
    'z': 'Diesel'
})


fuel_group = merged_data.groupby('fueltype').agg({
    'co2emissions': 'mean',
    'fuelconsumption_comb': 'mean',
    'model': 'count'
}).rename(columns={'model': 'vehicle_count'})

fuel_group = fuel_group.sort_values(by='co2emissions', ascending=False)
fuel_group
```

2. Average CO2 by Vehicle Class

```python
if 'vehicleclass' in merged_data.columns:
    class_group = merged_data.groupby('vehicleclass').agg({
        'co2emissions': 'mean',
        'fuelconsumption_comb': 'mean',
        'model': 'count'
    }).rename(columns={'model': 'vehicle_count'})

class_group.sort_values(by='co2emissions', ascending=False)
```

3. Top CO2 Emitters

```python
top_emitters = merged_data.groupby(['make', 'model'])['co2emissions'].max().reset_index()
top_emitters = top_emitters.sort_values(by='co2emissions', ascending=False).head(10)

# Reset index starting from 1
top_emitters.reset_index(drop = True, inplace = True)
top_emitters.index += 1
top_emitters.index.name = 'No.'
top_emitters
```

4. Emission trends by mechanical specs.

```python
merged_data.groupby('cylinders')['co2emissions'].mean()
merged_data.groupby('enginesize')['co2emissions'].mean()
merged_data
```

5. Most emisssions based on states

```python
# Group by 'state' and aggregate
state_summary = merged_data.groupby('state').agg({
    'co2emissions': ['mean', 'max', 'count']
}).sort_values(('co2emissions', 'mean'), ascending=False)

# Reset index to turn 'state' into a regular column
state_summary = state_summary.reset_index()

# Add an incremented index starting from 1
state_summary.index += 1
state_summary.index.name = 'No.'

# Display the result
state_summary
```

6.  Relationship between fuel consumption and emissions

```python
merged_data['fueltype'] = merged_data['fueltype'].replace({
    'x': 'Petrol',
    'z': 'Diesel'
})
merged_data['fuel_efficiency'] = merged_data['co2emissions'] / merged_data['fuelconsumption_comb']

merged_data.groupby('fueltype')['fuel_efficiency'].mean()
```
